```python
import diffusers
import torch

print("Diffusers version:", diffusers.__version__)
print("Torch version:", torch.__version__)
print("Packages loaded successfully!")
```

```python
from diffusers import StableDiffusionPipeline
import torch

model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float32
)

pipe = pipe.to("cpu")

print("Image generation model loaded successfully!")
```

```python
prompt = """
A modern IKEA-style living room, Scandinavian interior design,
comfortable sofa, wooden coffee table, indoor plants,
warm natural lighting, clean minimal decoration,
realistic interior photography
"""

image = pipe(prompt).images[0]

display(image)
```

```python
image.save("../generated_room.png")

print("Image saved successfully!")
```
