```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('../data/ikea.csv')

print("Shape:", df.shape)
print(df.head())
```

```python
# Create cleaned IKEA dataset

df = df.drop_duplicates(subset=["item_id"])

df = df.dropna(
    subset=["name", "category", "price", "short_description"]
)

# Create text content for RAG
df["content"] = df.apply(
    lambda x: (
        f"Product Name: {x['name']}. "
        f"Category: {x['category']}. "
        f"Price: {x['price']}. "
        f"Description: {x['short_description']}. "
        f"Designer: {x['designer']}. "
        f"Dimensions: {x['width']}cm width x "
        f"{x['depth']}cm depth x "
        f"{x['height']}cm height."
    ),
    axis=1
)

# Save cleaned dataset
df.to_csv(
    "../data/cleaned_ikea.csv",
    index=False
)

print("Cleaning completed!")
print("Shape:", df.shape)
print("File saved: ../data/cleaned_ikea.csv")
print("\nColumns:")
print(df.columns.tolist())
```
