```python
import pandas as pd

reviews = pd.read_csv("../data/reviews.csv")

print("Reviews loaded successfully!")
print("Shape:", reviews.shape)
print(reviews.head())
```

```python
from textblob import TextBlob

def get_sentiment(text):
    polarity = TextBlob(str(text)).sentiment.polarity
    
    if polarity > 0.1:
        return "Positive"
    elif polarity < -0.1:
        return "Negative"
    else:
        return "Neutral"

reviews["sentiment"] = reviews["review"].apply(get_sentiment)

print("Sentiment analysis completed!")
print(reviews[["review", "sentiment"]].head(10))
```

```python
sentiment_summary = reviews["sentiment"].value_counts()

print("Sentiment Summary:")
print(sentiment_summary)
```

```python
import matplotlib.pyplot as plt

sentiment_summary.plot(kind="bar")

plt.title("IKEA Customer Review Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.show()
```
